```python
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/ltv_dataset.csv")

df.head()
```

```python
#Remove columns that should not be used for prediction.

X = df.drop(columns=["LTV", "customerID"])

X.head()
```

```python
#Load the Trained Churn Model
model = joblib.load("../models/churn_model.pkl")

model
```

```python
#Create the SHAP Explainer
explainer = shap.Explainer(model)
```

```python
#Generate SHAP Values

#To make computation faster, use a sample of the data.

sample = X.sample(200, random_state=42)

shap_values = explainer(sample)
```

```python
#Create the SHAP Beeswarm Summary Plot
shap.plots.beeswarm(shap_values)
```

```python
#Save the Beeswarm Plot
plt.savefig(
    "../docs/images/shap_summary_beeswarm.png",
    bbox_inches="tight"
)

plt.close()
```

```python
#Create the SHAP Bar Plot
shap.plots.bar(shap_values)
```

```python
#Save the Bar Plot
plt.savefig(
    "../docs/images/shap_summary_bar.png",
    bbox_inches="tight"
)

plt.close()
```

```python
## Day 16 Conclusion

# The SHAP summary plots provide a global explanation of the churn model.

# Key observations:

# - Features at the top have the greatest influence on predictions.
# - The beeswarm plot shows both feature importance and the direction of impact.
# - The bar plot ranks features by their average contribution to the model.

# These visualizations help explain why the model predicts churn for different customers.
```
