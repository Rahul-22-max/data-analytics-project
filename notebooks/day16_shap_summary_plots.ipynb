```python
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
```

```python
df = pd.read_csv("../Data/processed/ltv_dataset.csv")

df.head()
```

```python
#Remove columns that should not be used for prediction.

X = df.drop(columns=["LTV", "customerID"])

X.head()
```

```python
#Load the Trained Churn Model
model = joblib.load("../models/churn_model.pkl")

model
```

```python
#Create the SHAP Explainer
raw_df = pd.read_csv("../Data/raw/Telco-Customer-Churn.csv")
raw_df["TotalCharges"] = pd.to_numeric(raw_df["TotalCharges"], errors="coerce").fillna(0)
X_model = pd.DataFrame(index=raw_df.index)
X_model["gender"] = (raw_df["gender"].eq("Female")).astype(int)
for col in ["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]:
    X_model[col] = raw_df[col].astype(float)
for col, value in [("Partner", "Yes"), ("Dependents", "Yes"), ("PhoneService", "Yes"), ("PaperlessBilling", "Yes")]:
    X_model[f"{col}_Yes"] = (raw_df[col].eq(value)).astype(int)
for col, values in [
    ("MultipleLines", {"No phone service": "MultipleLines_No phone service", "Yes": "MultipleLines_Yes"}),
    ("InternetService", {"Fiber optic": "InternetService_Fiber optic", "No": "InternetService_No"}),
    ("OnlineSecurity", {"No internet service": "OnlineSecurity_No internet service", "Yes": "OnlineSecurity_Yes"}),
    ("OnlineBackup", {"No internet service": "OnlineBackup_No internet service", "Yes": "OnlineBackup_Yes"}),
    ("DeviceProtection", {"No internet service": "DeviceProtection_No internet service", "Yes": "DeviceProtection_Yes"}),
    ("TechSupport", {"No internet service": "TechSupport_No internet service", "Yes": "TechSupport_Yes"}),
    ("StreamingTV", {"No internet service": "StreamingTV_No internet service", "Yes": "StreamingTV_Yes"}),
    ("StreamingMovies", {"No internet service": "StreamingMovies_No internet service", "Yes": "StreamingMovies_Yes"}),
    ("Contract", {"One year": "Contract_One year", "Two year": "Contract_Two year"}),
    ("PaymentMethod", {"Credit card (automatic)": "PaymentMethod_Credit card (automatic)", "Electronic check": "PaymentMethod_Electronic check", "Mailed check": "PaymentMethod_Mailed check"}),
]:
    for src, target in values.items():
        X_model[target] = (raw_df[col].eq(src)).astype(int)
expected = list(model.feature_names_in_)
for col in expected:
    if col not in X_model.columns:
        X_model[col] = 0
X_model = X_model[expected].astype(float)
sample_idx = np.random.RandomState(42).choice(
    X_model.shape[0], size=min(200, X_model.shape[0]), replace=False
)
sample = X_model.iloc[sample_idx]
explainer = shap.LinearExplainer(model, sample)
```

```python
#Generate SHAP Values

#To make computation faster, use a sample of the data.

shap_values = explainer(sample)
shap_values.values = np.asarray(shap_values.values, dtype=float)
```

```python
#Create the SHAP Beeswarm Summary Plot
shap.plots.beeswarm(shap_values)
```

```python
#Save the Beeswarm Plot
output_dir = Path("../docs/images")
output_dir.mkdir(parents=True, exist_ok=True)
plt.savefig(
    output_dir / "shap_summary_beeswarm.png",
    bbox_inches="tight"
)

plt.close()
```

```python
#Create the SHAP Bar Plot
shap.plots.bar(shap_values)
```

```python
#Save the Bar Plot
output_dir = Path("../docs/images")
output_dir.mkdir(parents=True, exist_ok=True)
plt.savefig(
    output_dir / "shap_summary_bar.png",
    bbox_inches="tight"
)

plt.close()
```
