```python
#Load Dataset

import pandas as pd

df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")
```

```python
#Clean TotalCharges
#Convert to numeric:

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

#Check missing values:

df.isnull().sum()

#Remove rows with missing values:

df = df.dropna()

#Verify:

df.shape
```

```python
#Generate LTV Column
#Create the baseline formula:

df["LTV"] = (
    df["MonthlyCharges"]
    * df["tenure"]
)

#Preview:

df[
    [
        "MonthlyCharges",
        "tenure",
        "LTV"
    ]
].head()
```

```python
#Validate LTV Values
#Check summary statistics:

df["LTV"].describe()

#Check for negative values:

(df["LTV"] < 0).sum()
```

```python
#Select Modeling Columns
#Create a cleaner dataset.

selected_columns = [
    "customerID",
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "tenure",
    "PhoneService",
    "InternetService",
    "Contract",
    "MonthlyCharges",
    "TotalCharges",
    "Churn",
    "LTV"
]

#Create dataset:

ltv_df = df[selected_columns]

#Check:

ltv_df.head()
```

```python
#Verify Dataset
#Shape:

ltv_df.shape

#Column names:

ltv_df.columns

#Data types:

ltv_df.info()
```

```python
#Save ltv_dataset.csv

ltv_df.to_csv(
    "../data/processed/ltv_dataset.csv",
    index=False
)
```
