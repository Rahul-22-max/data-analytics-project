```python
#Load Dataset

import pandas as pd
import numpy as np

df = pd.read_csv("../data/Telco-Customer-Churn.csv")
```

```python
#Convert TotalCharges:

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

df = df.dropna()
```

```python
#Review Revenue Columns
#Check:

df[
    [
        "tenure",
        "MonthlyCharges",
        "TotalCharges"
    ]
].describe()
```

```python
#Analyze Option 1 — TotalCharges

# Pros
# Already exists
# Easy regression target
# Real revenue
# Cons
# Historical revenue only
# Not future value
# Doesn't predict future customer worth
```

```python
df["TotalCharges"].describe()
```

```python
# Analyze Option 2 — Estimated Future Revenue
# Pros
# Future-focused
# Better business value
# Cons
# Assumes all customers stay
# Ignores churn risk
```

```python
remaining_months = 12

df["FutureRevenue"] = (
    df["MonthlyCharges"]
    * remaining_months
)
```

```python
df["FutureRevenue"].head()
```

```python
#Analyze Option 3 — Custom LTV Formula
```

```python
df["EstimatedLTV"] = (
    df["MonthlyCharges"]
    * df["tenure"]
)
```

```python
df["EstimatedLTV"].describe()
```

```python
#Improve the Formula
#Create a retention probability.
```

```python
#Temporary example:
df["RetentionProb"] = np.where(
    df["Churn"] == "No",
    0.9,
    0.3
)
```

```python
#Enhanced LTV:
df["EnhancedLTV"] = (
    df["MonthlyCharges"]
    * df["tenure"]
    * df["RetentionProb"]
)
```

```python
#View results:
df[
    [
        "MonthlyCharges",
        "tenure",
        "RetentionProb",
        "EnhancedLTV"
    ]
].head()
```
