```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/ltv_dataset.csv")
df.head()

df.shape

df.info()
```

```python
#Define Features and Target
target = "LTV"

#Create X and y:

X = df.drop(columns=["LTV"])
y = df["LTV"]

#Remove customerID because it is an identifier, not a predictive feature:

X = X.drop(columns=["customerID"])
```

```python
#Identify Numerical and Categorical Columns
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "SeniorCitizen"
]

#Categorical columns
categorical_features = [
    col for col in X.columns if col not in numerical_features
]

#Check them:

print("Numerical Features:", numerical_features)
print("Categorical Features:", categorical_features)
```

```python
#Create Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

#Check shapes:

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#Build the Preprocessing Pipeline
#Numeric pipeline
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

#Categorical pipeline
categorical_transformer = Pipeline(
    steps=[
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

#Combine both using ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
#Create Full Linear Regression Pipeline
#Now combine preprocessing + model into one pipeline.

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)
```

```python
#Train the Linear Regression Model
linear_model.fit(X_train, y_train)
```

```python
#Make Predictions
#On training data
y_train_pred = linear_model.predict(X_train)

#On test data
y_test_pred = linear_model.predict(X_test)

#Check a few predictions:

y_test_pred[:10]
```

```python
#Evaluate the Model

#Now calculate regression metrics.

#1) MAE — Mean Absolute Error

mae = mean_absolute_error(y_test, y_test_pred)
print("MAE:", mae)

#2) MSE — Mean Squared Error

mse = mean_squared_error(y_test, y_test_pred)
print("MSE:", mse)

#3) RMSE — Root Mean Squared Error

rmse = np.sqrt(mse)
print("RMSE:", rmse)

#4) R² Score

r2 = r2_score(y_test, y_test_pred)
print("R2 Score:", r2)
```

```python
#Create a Results Table

#Compare actual vs predicted values:

results = pd.DataFrame({
    "Actual_LTV": y_test.values,
    "Predicted_LTV": y_test_pred
})

results.head(10)
```

```python
#Plot Actual vs Predicted

#This is one of the most important visualizations for regression.

plt.figure(figsize=(8,6))
sns.scatterplot(x=y_test, y=y_test_pred)
plt.xlabel("Actual LTV")
plt.ylabel("Predicted LTV")
plt.title("Actual vs Predicted LTV - Linear Regression")
plt.show()
```

```python
#Plot Residual Errors

#Residual = Actual - Predicted

residuals = y_test - y_test_pred

plt.figure(figsize=(8,6))
sns.histplot(residuals, bins=30, kde=True)
plt.title("Residual Error Distribution")
plt.xlabel("Error")
plt.show()
```

```python
#Save the Trained Model

#Create a models folder if it doesn’t exist:

#mkdir -p models

#Now save the model:

joblib.dump(linear_model, "../models/linear_regression_ltv.pkl")

#Test loading it back:

loaded_model = joblib.load("../models/linear_regression_ltv.pkl")
loaded_model
```
